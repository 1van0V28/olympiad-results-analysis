## Пропуски

```python
missing = pd.DataFrame({
	'missing_count': df.isna().sum(),
	'missing_fraction': df.isna().mean().round(2)
}).sort_values('missing_fraction', ascending=False)
display(missing[missing['missing_count'] > 0])
```

## Дубликаты

```python
print('Полных дупликатов:', df.duplicated().sum())

duplicates = df[df.duplicated(keep=False)].sort_values('student_id')
print('Всего строк:', len(duplicates))
display(duplicates)
```

## Числовые признаки

```python
task_n = df.columns[df.columns.str.match(r"^Task_\d+_time_spent$")].tolist()
numerical_characteristics = (
	['total_score', 'attempts_count', 'disconnects'] +
	task_n
)

display(df[numerical_characteristics].describe().T)
```

### Проверка расхождения 'start_time' и 'end_time'

```python
parsed_start_time = pd.to_datetime(df['start_time'], format='%Y-%m-%d %H:%M:%S', errors='coerce')
parsed_end_time = pd.to_datetime(df['end_time'], format='%Y-%m-%d %H:%M:%S', errors='coerce')
print('Ошибки \'start_time\':', parsed_start_time.isna().sum())
print('Ошибки \'end_time\':', parsed_start_time.isna().sum())

time_discrepancies = parsed_start_time >= parsed_end_time
print('Расхождения времени:', time_discrepancies.sum())
```

## Категориальные признаки

```python
category_columns = ['region', 'grade', 'device_type']

display(df[category_columns].nunique(dropna=False))

for column in category_columns:
	display(pd.DataFrame({
		'count': df[column].value_counts(dropna=False),
		'fraction': df[column].value_counts(normalize=True, dropna=False).round(2)
	}).sort_index())
```

### Проверка на возможные значения

```python
task_n_answer_columns = df.columns[df.columns.str.match(r"^Task_\d+_answer$")].tolist()
task_n_correct_columns = df.columns[df.columns.str.match(r"^Task_\d+_correct$")].tolist()

display(df[task_n_answer_columns + task_n_correct_columns].nunique(dropna=False))


def get_task_invalid_unique(task_columns, valid_values):
	return pd.DataFrame({
		task_n_valid: df.loc[
			~(df[task_n_valid].isin(valid_values) | df[task_n_valid].isna()), 
			task_n_valid
		].value_counts(dropna=False) for task_n_valid in task_columns
	}).T


valid_task_answer = [1, 2, 3, 4]
display(get_task_invalid_unique(task_n_answer_columns, valid_task_answer))

valid_task_correct = [0, 1]
display(get_task_invalid_unique(task_n_correct_columns, valid_task_correct))
```

### Проверка на несовпадение пропусков

```python
task_answer_isna = df[task_n_answer_columns].isna()
task_correct_isna = df[task_n_correct_columns].isna()

task_answer_isna.columns = [f'Task_{n}' for n in range(1, len(task_n_answer_columns) + 1)]
task_correct_isna.columns = [f'Task_{n}' for n in range(1, len(task_n_correct_columns) + 1)]
mismatch = task_answer_isna != task_correct_isna
display(mismatch.sum())
```

### Проверка времени пропущенного вопроса

```python
task_n_time_spent_columns = df.columns[df.columns.str.match(r"^Task_\d+_time_spent$")].to_list()
task_time_spent_isna = df[task_n_time_spent_columns].isna()

task_time_spent_isna.columns = [f'Task_{n}' for n in range(1, len(task_n_time_spent_columns) + 1)]

mismatch = task_time_spent_isna != task_correct_isna
display(mismatch.sum())
```

### Проверка на несоответствие результата

```python
score = df[task_n_correct_columns].sum(axis=1)
score_mismatch = ~(score == df['total_score'].astype(int))
print('Количество несоответствий результату:', score_mismatch.sum())
```
