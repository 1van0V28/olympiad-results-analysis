```python
import pandas as pd
import matplotlib as plt


df = pd.read_parquet('../data/olympiad_results_clean.parquet')
df.info(memory_usage=True)
```

```python
categories = ['region', 'grade', 'device_type']
for category in categories:
	category_values = df[category].value_counts(normalize=True).round(2)
	display(category_values)
```

```python
task_n = df.columns[df.columns.str.match(r"^Task_\d+_(correct|time_spent)$")].tolist()
numerical_characteristics = (
	['total_score', 'attempts_count', 'disconnects'] +
	task_n
)

display(df[numerical_characteristics].describe().T)
```
